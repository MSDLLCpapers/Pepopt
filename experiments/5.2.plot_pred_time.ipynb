```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
plt.rcParams.update({
    # 'font.family': 'serif',
    # 'font.serif': ['Times', 'Times', 'DejaVu Serif'],
    'font.size': 12,
    'font.weight': 'normal',
    'axes.titlesize': 12,
    'axes.titleweight': 'normal',
    'axes.labelsize': 12,
    'axes.labelweight': 'normal',
    'xtick.labelsize': 12,
    'ytick.labelsize': 12,
    'legend.fontsize': 12,
})
```

```python
# 两个折线图，

table_380 = pd.read_csv('results/pred_time/traject_380/best_scores.csv')
f1 = plt.figure(figsize=(8, 4))
# each row is a model, each column is a time stamp
for i in range(10):
    plt.plot(range(51), table_380.iloc[i].values, markersize=5, marker='o', label=f'Model {i+1}')

plt.xlabel('Epoch')
plt.ylabel('pEC50 (prediction)')

plt.hlines(11.576, 0, 50, colors='b', linestyles='dashed', label='Best score found by DP')
plt.text(0, 11.57, 'Best score found by DP', color='b', fontsize=12, va='top', ha='left')
plt.title('(a) GA search progress for used AAs space ($10^{20}$ sequences)')
plt.tight_layout()
plt.savefig('results/paper/hi-res_figures/si_ga_progress_380.png', dpi=300)
```
